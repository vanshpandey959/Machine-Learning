Series

A series is a one dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

Creating Series

```python
import numpy as np
import pandas as pd
```

```python
#labels also called index
labels = ['a','b','c']
my_list = [10,20,30];
arr = np.array([10,20,30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr) # only 1 dim
```

```python
pd.Series(d)
```

Missing Data

```python

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

Removing Missing Data

```python
df.dropna()
# Row with even a single null value will be deleted
```

```python
df.dropna(thresh=3)
# Atmost 3 non null honi chahiye
```

Filling the missing data

```python
df.fillna(0)
```

```python
values = {
    'A' : 0,
    'B': 100,
    'C': 300,
    'D': 400
}
df.fillna(value=values)
```

```python
# Drop happens row wise
# Selection and filling happens column wise
```

```python
df
```

```python
# Now I want to fill through mean of data
df.fillna(df.mean())
```

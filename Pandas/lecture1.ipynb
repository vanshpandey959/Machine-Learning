Series

A series is a one dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

Creating Series

```python
import numpy as np
import pandas as pd
```

```python
#labels also called index
labels = ['a','b','c']
my_list = [10,20,30];
arr = np.array([10,20,30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr) # only 1 dim
```

```python
pd.Series(d)
```

DataFrame - When we combine multiple series it forms Dataframe

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
pd.DataFrame(data)
```

Quantitive data  = Age, Salary
Qualitive data = Gender

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
db = sns.load_dataset('tips')
```

```python
db
```

Distribution Plots - To plot quantitive / numerical data 

```python
sns.displot(db['total_bill'], bins=20, kde=True)
```

Dist Plot - Divides in bins limme 10-12 , 12-14 ... and values are there in these bills

Kde = How much are data is skewed 

```python
plt.subplot(2,2,1)
sns.histplot(db['total_bill'], bins=20, kde=True)
plt.subplot(2,2,2)
sns.histplot(db['tip'], bins=20, kde=True)
```

```python
# Joint Plot
sns.jointplot(x='total_bill', y='tip', data=db, kind='scatter')
```

```python
sns.jointplot(x='total_bill', y='tip', data=db, kind='kde')
```

```python
sns.jointplot(x='total_bill', y='tip', data=db, kind='hex')
```

```python
sns.pairplot(db, hue='sex')
```

```python
sns.rugplot(db['total_bill'])
```

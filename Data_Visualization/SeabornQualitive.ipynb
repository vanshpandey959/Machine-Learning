Categorical Data Plots - boxplot, violinplot, stripplot, swarm, bar and countplot

```python
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
# In bar plot in y axis there is usually quantitive value while in x - categorical
sns.barplot(x=df['sex'], y=df['total_bill'])
```

```python
# By default the quantitive axis is aligned be np.mean. We can change the estimator also
import numpy as np
sns.barplot(x=df['sex'], y=df['total_bill'],estimator=np.sum)
```

```python
# Box plot - Made using 5 point summary
sns.boxplot(x=df['tip'], y=df['day'])
# We are taking qualitative at y, quantitive at x
```

```python
sns.violinplot(x=df['day'], y=df['tip'])
```

```python
sns.stripplot(x='day',y='tip',data=df)
```

```python
sns.swarmplot(x='day',y='tip',data=df)
```

```python
sns.violinplot(x='tip', y='day', data=df)
sns.swarmplot(x='tip', y='day', data=df)
```

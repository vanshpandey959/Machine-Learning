Categorical Data Plots - boxplot, violinplot, stripplot, swarm, bar and countplot

```python
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
# In bar plot in y axis there is usually quantitive value while in x - categorical
sns.barplot(x=df['sex'], y=df['total_bill'])
```

```python
# By default the quantitive axis is aligned be np.mean. We can change the estimator also
import numpy as np
sns.barplot(x=df['sex'], y=df['total_bill'],estimator=np.sum)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
flights = sns.load_dataset('flights')
```

```python
flights
```

```python
tips = sns.load_dataset('tips')
```

Heat Map

```python
# Heat map shows correlation among the features
tipsscorr = tips[['total_bill', 'tip', 'size']]
tipsscorr
```

```python
tipsscorr.corr()
```

```python
sns.heatmap(tipsscorr.corr(), annot=True)
```

Cluster Map

```python
sns.clustermap(tipsscorr.corr())
```

```python
# pivot table Heat map
pvtflights = flights.pivot_table(values='passengers', index='month', columns='year')
```

```python
pvtflights
```

```python
sns.heatmap(pvtflights)
```

Regression Plot

```python
sns.lmplot(x='total_bill', y='tip', data=tips)
```

```python
sns.lmplot(x='total_bill', y='tip', data=tips, hue='sex')
```

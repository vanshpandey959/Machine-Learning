```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
import pandas as pd
```

```python
x = np.linspace(0, 5, 11)
x
```

```python
y = x ** 2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# By default a line plot
```

```python
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,x)
plt.subplot(2,2,3)
plt.plot(y,y)
```

Object Oriented Way

Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword argument.

figsize is a tuple of the width and height of the figure in inches, dpi is the dots-per-inch (pixel per inch)

```python
# These plots are designed on canvas, if we need full control over these canvas we use figure()
fig = plt.figure(figsize=(12,8),dpi=100)
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)

# The plots are overlapping. We can manipulate them accordingly, rather in subplot it was dividing by itself
```

More the dpi -  More clear visuals

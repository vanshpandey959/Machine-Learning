```python
# Eda Analysis
# We will use : FuelType, Mileage, EngineSize, year , tax, 
# Transmission is disputed (Can and can not use bcs all three types have approx same prices)
```

```python
X = df.drop(columns=['price'])
X
```

```python
y = df['price']
```

```python
# Convert categorical data to numerical
df.columns
```

```python
# Model, transmission and fueltype is categorical
# Doing One hot encoding
X_one_code = pd.get_dummies(X, columns = ['model', 'fuelType', 'transmission'], drop_first=True)
```

```python
X_one_code = X_one_code.astype(int)
X_one_code
```

```python
# Lets also use Label Encoding
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
columns = ['model', 'fuelType', 'transmission']
Xlabel = X.copy()
label_encoders = {}

for col in columns:
    le = LabelEncoder()
    Xlabel[col] = le.fit_transform(Xlabel[col].astype(str))
    label_encoders[col] = le
```

```python
Xlabel
```

```python
Xlabel['model'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numerical_cols = ['year','mileage', 'tax', 'mpg', 'engineSize']
scaler = StandardScaler()
X_one_code[numerical_cols] = scaler.fit_transform(X_one_code[numerical_cols])
```

```python
X_one_code
```

```python
Xlabel[Xlabel.columns] = scaler.fit_transform(Xlabel[Xlabel.columns])
```

```python
Xlabel
```

```python
# Now lets move to the Model Creation Part
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_code, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
y_test
```

```python
r2 = r2_score(y_test, y_pred)
r2
```

```python
n = X_test.shape[0] # No of rows/data (test)
p = X_test.shape[1] # No. of columns

adjusted_r2 = 1 - ((1-r2) * (n-1) / (n-p-1))
adjusted_r2
```

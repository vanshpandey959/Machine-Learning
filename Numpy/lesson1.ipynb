Deep and Shallow Copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice * 10
```

```python
slice
```

```python
a
# No changes in a if changes in its slice
# This refers to Shallow Copy
```

```python
b = a
```

```python
b[0] = 99
```

```python
b
```

```python
a
```

```python
# Here a also gets changed when we change in b
# When we fully copy an array to another variable , the location is passed to other variable, so changes in b happens in a also

# We need to create a shallow copy
c = b.copy()
```

```python
c[0] = 100
c
```

```python
b 
# No change in b
```
